```python
import inspect

def Print(*args):
    # Get caller’s frame
    frame = inspect.currentframe().f_back
    # Get the source code line where Print was called
    call_line = inspect.getframeinfo(frame).code_context[0].strip()

    # Example: call_line == 'Print(i, j)'
    # Extract inside of parentheses
    inside = call_line[call_line.find('(')+1:call_line.rfind(')')]

    # Split by commas to get variable names
    names = [name.strip() for name in inside.split(',')]

    for name, value in zip(names, args):
        print(f"[{type(value).__name__}]{name}={value}", end=", ")
    print()

for i in range(2):
    for j in range(2):
        Print(i+1, j)
```
