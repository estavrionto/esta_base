```python
import numpy as np
import matplotlib.pyplot as plt
import copy
start = {
    'x_totals' : [1,1,1,1,],
    'y_totals' : [0,4,0,],
    'track' : [
        {
            'x':0,
            'y':1,
            'cons':['right','left'],
            # 'cons':['right','top'],
            # 'status':'given',
        },
            {
            'x':3,
            'y':1,
            'cons':['right','left'],
            # 'cons':['left','bot'],

            # 'status':'given',
        },
    ],
}

start = {
    'x_totals' : [4,4,2,3,3,3],
    'y_totals' : [4,2,1,3,5,4],
    'track' : [
        {
            'x':4,
            'y':0,
            'cons':['bot','left'],
            # 'cons':['right','top'],
            # 'status':'given',
        },
            {
            'x':5,
            'y':5,
            'cons':['bot','right'],
            # 'cons':['left','bot'],

            # 'status':'given',
        },
    ],
}

            
def prepare_grid(start):
    t = {'cons':{
        'left':None,
        'right':None,
        'top':None,
        'bot':None,
    },
         'status':None,}
    grid = [[copy.deepcopy(t) for j in range(len(start['x_totals']))] for i in range(len(start['y_totals']))]
    # print(grid)
    for i in range(len(start['y_totals'])):
        for j in range(len(start['x_totals'])):
            # print(i,j)
            for k in start['track']:
                if k['x']==j:
                    if k['y']==i:
                        tt = copy.deepcopy(t['cons'])
                        for p in k['cons']:
                            tt[p]=True
                        grid[i][j]={'cons':tt,'status':'given',}
                        

    for i in grid:
        print(i)
    return {
        'x_totals':start['x_totals'],
        'y_totals':start['y_totals'],
        'track_grid':grid,
        }
                
        
# grid_init = {
#     start['x']
# }

grid_init = prepare_grid(start)
grid_init
```

```python

def plot_grid(grid): 
    # x = np.random.randint(1, 14, 50)
    # y = np.random.randint(1, 9, 50)
    
    # plt.plot(x, y, 'bo')
    # Create the figure and axes objects explicitly
    fig, ax = plt.subplots(figsize=(5, 5))
    len_x = len(grid['x_totals'])
    len_y = len(grid['y_totals'])
    vacant_x = [i for i in grid['x_totals']]
    vacant_y = [i for i in grid['y_totals']]
    s = 0.5
    s2 = 0.3
    color1 = 'black'
    color2 = 'blue'

    for i,m in enumerate(grid['x_totals']):
        # print(i,m)
        ax.text(x=i+s, y=len_y+s, s=m, fontsize=12, color=color1, va="center", ha="center")

    for i,m in enumerate(grid['y_totals']):
        # print(i,m)
        ax.text(x=len_x+s, y=i+s, s=m, fontsize=12, color=color1, va="center", ha="center")

    for x in range(len_x):
        for y in range(len_y):
            val = grid['track_grid'][y][x]
            # print(f'x={x},y={y}',val)
            if val['status']!=None:
                if val['status'] == 'given':
                    color = color1
                else:
                    color = color2
                ax.plot(x+s, y+s, color=color, marker="o", linestyle="None")
                # plt.plot(x+s, y+s, 'bo')
                vacant_x[x] = vacant_x[x]-1
                vacant_y[y] = vacant_y[y]-1
                for dire in val['cons']:
                    if val['cons'][dire]==True:
                    # print(dire)
                        if dire == 'left':
                            ax.plot(x+s-s2, y+s, color=color, marker="o", linestyle="None")
                        if dire == 'right':
                            ax.plot(x+s+s2, y+s, color=color, marker="o", linestyle="None")
                        if dire == 'top':
                            ax.plot(x+s, y+s+s2, color=color, marker="o", linestyle="None")
                        if dire == 'bot':
                            ax.plot(x+s, y+s-s2, color=color, marker="o", linestyle="None")

    # for i,m in enumerate(vacant_x):
    #     # print(i,m)
    #     ax.text(x=i+s, y=len_y+1+s, s=m, fontsize=12, color=color1, va="center", ha="center")

    # for i,m in enumerate(vacant_y):
    #     # print(i,m)
    #     ax.text(x=len_x+1+s, y=i+s, s=m, fontsize=12, color=color1, va="center", ha="center")

    
    # Apply the ticks and grid directly to the ax object
    ax.set_xticks(range(3 + len_x))
    ax.set_yticks(range(3 + len_y))
    ax.grid(linestyle='dotted')
    ax.tick_params(axis="x", labelbottom=False)
    ax.tick_params(axis="y", labelleft=False)
    return sum(vacant_x)+sum(vacant_y)
plot_grid(grid=grid_init)
```

```python
def solve_1s_propagate1(grid):
    grid = copy.deepcopy(grid)
    len_x = len(grid['x_totals'])
    len_y = len(grid['y_totals'])
    l_x = [i for i in range(len_x)]
    l_y = [i for i in range(len_y)]
    for x in l_x:
        for y in l_y:
            val = grid['track_grid'][y][x]
            # print(f'x={x},y={y}',val)
            if val['status'] in ['given','solved']:
                for dire in val['cons']:
                    if dire == 'left':
                        xn=x-1
                        yn=y
                        diren='right'
                    if dire == 'right':
                        xn=x+1
                        yn=y
                        diren='left'
                    if dire == 'top':
                        xn=x
                        yn=y+1
                        diren='bot'
                    if dire == 'bot':
                        xn=x
                        yn=y-1
                        diren='top'
                    if xn in l_x and yn in l_y:
                        # print(val,dire,xn,yn,diren)
                        if diren not in grid['track_grid'][yn][xn]['cons']:
                            grid['track_grid'][yn][xn]['cons'].append(diren)
                            if grid['track_grid'][yn][xn]['status']==None:
                                grid['track_grid'][yn][xn]['status']='fill1'
                            if grid['track_grid'][yn][xn]['status']=='fill1':
                                grid['track_grid'][yn][xn]['status']='solved'


    return grid
    
def solve_1s_solve1(grid):
    grid = copy.deepcopy(grid)
    return grid

grid = copy.deepcopy(grid_init)
for i in range(10):
    score = plot_grid(grid=grid)
    if score==0:
        break
    grid = solve_1s_propagate1(grid)
    grid = solve_1s_solve1(grid)

```
