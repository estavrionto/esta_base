```python
import i3ipc

i3 = i3ipc.Connection()

def on_window_focus(i3conn, e):
    wm_class = e.container.window_class
    title = e.container.name
    if wm_class == "spring":
        print(f"[ON] spring focused — title='{title}'")
    else:
        print(f"[OFF] focus -> {wm_class} | {title}")

# initial focus
focused = i3.get_tree().find_focused()
if focused:
    print("Initial focus:", focused.window_class, "|", focused.name)

i3.on("window::focus", on_window_focus)
print("Subscribed. Change focus to see events…")
i3.main()
```

```python
import time
import i3ipc
from Xlib import X, display

# set this to the pixel location **inside** the spring window (relative offset)
OFF_X, OFF_Y = 1, 1   # <-- change these later
OFF_X, OFF_Y = 333, 65   # <-- change these later
OFF_X, OFF_Y = 545, 65

# detect this in spring windows class: RGB@(333,65) = (5, 30, 46)
# my script is at (base) [ab@archlinux ~]$ ~/Software/esta_base/BAR/bar_cover.sh 


i3 = i3ipc.Connection()
d  = display.Display()
root = d.screen().root

def get_focused():
    return i3.get_tree().find_focused()

def sample_rgb_at_screen(x, y):
    img = root.get_image(x, y, 1, 1, X.ZPixmap, 0xffffffff)
    # Xorg usually returns BGRA bytes; take first 3 channels
    b = img.data[0]; g = img.data[1]; r = img.data[2]
    return (r, g, b)

print("Sampling RGB at (window + offset) while 'spring' is focused. Ctrl+C to stop.")
last = None
try:
    while True:
        f = get_focused()
        if f and f.window_class == "spring":
            ax = f.rect.x + OFF_X
            ay = f.rect.y + OFF_Y
            rgb = sample_rgb_at_screen(ax, ay)
            if rgb != last:
                print(f"RGB@({ax},{ay}) = {rgb}")
                last = rgb
        time.sleep(0.1)  # 10 Hz
except KeyboardInterrupt:
    pass
# RGB@(545,65) = (6, 34, 52) ingame, normal
# RGB@(545,65) = (4, 68, 130) ingame, highlight
# RGB@(545,65) = (5, 28, 44) nogame, normal
# RGB@(545,65) = (3, 66, 129) nogame, highlight
# RGB@(545,65) = (3, 17, 26) ingame, leave
# RGB@(545,65) = (2, 14, 22) nogame, exit
```

```python
#!/usr/bin/env python3
import os, signal, time
import i3ipc
from pathlib import Path
from Xlib import X, display

# ---- CONFIG ----
TARGET_CLASS   = "spring"
SAMPLE_HZ      = 60.0
PATCH_SCRIPT   = str(Path.home() / "Software/esta_base/BAR/bar_cover.sh")
PIDFILE        = Path("/tmp/screen_patches.pid")
VERBOSE        = False  # set True to log checks
# Pixels (window-relative)
GATE_POS       = (0, 0)
OFF_CHECK_POS  = (545, 65)
OFF_COLORS     = {(0, 0, 0), (27, 27, 28), (3, 3, 3), (24, 24, 25)}  # any of these -> OFF
OFF_COLORS     = {(0, 0, 0), (27, 27, 28), (33, 33, 33)}  # any of these -> OFF
MIN_OFF_SEC    = 1.0   # delay before turning OFF once OFF is detected
# ----------------

def start_patches():
    if PIDFILE.exists() and PIDFILE.stat().st_size > 0:
        return
    os.spawnlp(os.P_NOWAIT, "bash", "bash", "-c", f'"{PATCH_SCRIPT}" >/dev/null 2>&1')

def stop_patches():
    if PIDFILE.exists():
        for line in PIDFILE.read_text().splitlines():
            try: os.kill(int(line.strip()), signal.SIGTERM)
            except Exception: pass
        try: PIDFILE.unlink()
        except Exception: pass

def sample_rgb(root, x, y):
    img = root.get_image(x, y, 1, 1, X.ZPixmap, 0xffffffff)
    b = img.data[0]; g = img.data[1]; r = img.data[2]  # BGRA -> RGB
    return (r, g, b)

def visible_spring_windows(i3):
    """Return list of 'spring' containers on currently visible workspaces."""
    visible_ws = {ws.name for ws in i3.get_workspaces() if ws.visible}
    out = []
    for con in i3.get_tree().leaves():
        if not con.window: continue
        if con.window_class != TARGET_CLASS: continue
        ws = con.workspace()
        if ws and ws.name in visible_ws:
            out.append(con)
    return out

def should_turn_on(i3, root):
    """
    Gate: (0,0) must be (0,0,0). If gate passes AND (545,65) not in OFF_COLORS -> ON.
    Returns: (on_now: bool, probe_info: (ax,ay,ox,oy,off_rgb) or None)
    """
    cons = visible_spring_windows(i3)
    last_probe_info = None

    for con in cons:
        # Gate check at (0,0)
        gax, gay = con.rect.x + GATE_POS[0], con.rect.y + GATE_POS[1]
        try:
            gate_rgb = sample_rgb(root, gax, gay)
        except Exception:
            continue

        if VERBOSE:
            print(f"[gate] ws={con.workspace().name} rgb@{GATE_POS}={gate_rgb}")

        if gate_rgb != (0, 0, 0):
            continue  # strict OFF for this window

        # Off-check at (545,65)
        ax, ay = con.rect.x + OFF_CHECK_POS[0], con.rect.y + OFF_CHECK_POS[1]
        try:
            off_rgb = sample_rgb(root, ax, ay)
        except Exception:
            continue

        if VERBOSE:
            print(f"[check] rgb@{OFF_CHECK_POS}={off_rgb} off_colors={OFF_COLORS}")

        last_probe_info = (ax, ay, OFF_CHECK_POS[0], OFF_CHECK_POS[1], off_rgb)

        if off_rgb in OFF_COLORS:
            continue  # still OFF for this window; try next

        # Gate passed and off-check not in OFF_COLORS => ON
        return True, last_probe_info

    return False, last_probe_info

# ---- main loop ----
i3   = i3ipc.Connection()
dpy  = display.Display()
root = dpy.screen().root

period      = 1.0 / SAMPLE_HZ
patches_on  = False
off_at      = None  # monotonic timestamp when we're allowed to turn OFF (None = no pending off)
print(f"Watching visible 'spring' at {SAMPLE_HZ:.0f} Hz (gate + off-colors, 1s OFF delay). Ctrl+C to stop.")

try:
    while True:
        on_now, probe = should_turn_on(i3, root)
        now = time.monotonic()

        if on_now:
            # Immediate ON; clear any pending OFF timer
            if not patches_on:
                print("state -> ON")
                if probe:
                    ax, ay, ox, oy, off_rgb = probe
                    print(f"last probe: abs=({ax},{ay}) rel=({ox},{oy}) rgb={off_rgb}")
                start_patches()
                patches_on = True
            off_at = None  # reset OFF timer since we saw ON
        else:
            # No ON signal: arm/extend OFF timer
            if patches_on:
                if off_at is None:
                    off_at = now + MIN_OFF_SEC
                # If timer expired, turn OFF
                if now >= off_at:
                    print("state -> OFF")
                    if probe:
                        ax, ay, ox, oy, off_rgb = probe
                        print(f"last probe: abs=({ax},{ay}) rel=({ox},{oy}) rgb={off_rgb}")
                    stop_patches()
                    patches_on = False
                    off_at = None  # clear timer

        time.sleep(period)
except KeyboardInterrupt:
    stop_patches()
    print("Stopped and patches killed.")
```

```python
import os, signal, time
import i3ipc
from pathlib import Path
from Xlib import X, display

# ---- CONFIG ----
WATCHPOINTS = {
    # name: (offset_x, offset_y, target_rgb)
    # "lobby1": (333, 65, (5, 30, 46)),   # first pixel + target
    # "lobby2a": (545, 65, (5, 28, 44)),  # others
    # "lobby2b": (545, 65, (6, 34, 52)),
    # "lobby2c": (545, 65, (3, 66, 129)),
    # "lobby2d": (545, 65, (4, 68, 130)),

    # GATE (must match or patches NEVER turn on):
    "out_of_game": (0, 0, (0, 0, 0)),

    # ANY-OF (only considered if gate matches):
    "ingame_normal":    (545, 65, (6, 34,  52)),
    "ingame_highlight": (545, 65, (4, 68, 130)),
    "ingame_leave":     (545, 65, (3, 17,  26)),
    "nogame_normal":    (545, 65, (5, 28,  44)),
    "nogame_highlight": (545, 65, (3, 66, 129)),
    "nogame_exit":      (545, 65, (2, 14,  22)),
    
}

SAMPLE_HZ   = 60.0
PATCH_SCRIPT = str(Path.home() / "Software/esta_base/BAR/bar_cover.sh")
PIDFILE      = Path("/tmp/screen_patches.pid")
TARGET_CLASS = "spring"
VERBOSE      = False  # set True to log every probe
# ----------------

def start_patches():
    if PIDFILE.exists() and PIDFILE.stat().st_size > 0:
        return
    os.spawnlp(os.P_NOWAIT, "bash", "bash", "-c", f'"{PATCH_SCRIPT}" >/dev/null 2>&1')

def stop_patches():
    if PIDFILE.exists():
        for line in PIDFILE.read_text().splitlines():
            try: os.kill(int(line.strip()), signal.SIGTERM)
            except Exception: pass
        try: PIDFILE.unlink()
        except Exception: pass

def sample_rgb(root, x, y):
    img = root.get_image(x, y, 1, 1, X.ZPixmap, 0xffffffff)
    b = img.data[0]; g = img.data[1]; r = img.data[2]   # BGRA -> RGB
    return (r, g, b)

def visible_spring_windows(i3):
    """Return list of 'spring' containers on currently visible workspaces."""
    visible_ws = {ws.name for ws in i3.get_workspaces() if ws.visible}
    out = []
    for con in i3.get_tree().leaves():
        if not con.window: continue
        if con.window_class != TARGET_CLASS: continue
        ws = con.workspace()
        if ws and ws.name in visible_ws:
            out.append(con)
    return out

# ---- main loop ----
i3   = i3ipc.Connection()
dpy  = display.Display()
root = dpy.screen().root

print(f"Watching gate + {len(WATCHPOINTS)-1} condition(s) on any VISIBLE 'spring' at {SAMPLE_HZ:.0f} Hz. Ctrl+C to stop.")
period     = 1.0 / SAMPLE_HZ
patches_on = False

try:
    while True:
        cons = visible_spring_windows(i3)
        on_now = False
        matched_key = None
        last_probe_info = None

        if cons:
            # evaluate per visible window: gate must match on the SAME window
            gx, gy, gtgt = WATCHPOINTS["out_of_game"]
            for con in cons:
                # 1) gate
                gax = con.rect.x + gx
                gay = con.rect.y + gy
                try:
                    grgb = sample_rgb(root, gax, gay)
                except Exception:
                    continue

                if VERBOSE:
                    print(f"[gate] ws={con.workspace().name} abs=({gax},{gay}) rgb={grgb} target={gtgt}")

                if grgb != gtgt:
                    # gate failed on this window; try next visible window
                    continue

                # 2) any-of (only if gate matched)
                for key, (ox, oy, tgt) in WATCHPOINTS.items():
                    if key == "out_of_game":
                        continue
                    ax = con.rect.x + ox
                    ay = con.rect.y + oy
                    try:
                        rgb = sample_rgb(root, ax, ay)
                        last_probe_info = (key, ax, ay, ox, oy, rgb)
                        if VERBOSE:
                            print(f"[check] {key}: abs=({ax},{ay}) rel=({ox},{oy}) rgb={rgb} target={tgt}")
                        if rgb == tgt:
                            matched_key = key
                            on_now = True
                            break
                    except Exception:
                        pass
                if on_now:
                    break  # no need to check other windows

        # Toggle immediately (no grace window)
        if on_now and not patches_on:
            print(f"state -> ON (matched={matched_key})")
            start_patches()
            patches_on = True
        elif not on_now and patches_on:
            print("state -> OFF")
            if last_probe_info:
                key, ax, ay, ox, oy, rgb = last_probe_info
                print(f"last probe: key={key} abs=({ax},{ay}) rel=({ox},{oy}) rgb={rgb}")
            stop_patches()
            patches_on = False

        time.sleep(period)
except KeyboardInterrupt:
    stop_patches()
    print("Stopped and patches killed.")
```
